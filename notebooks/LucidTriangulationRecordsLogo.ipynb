```python
# 1. Install required library
!pip install svgwrite -q

import math
import svgwrite
from google.colab import files


def create_lucid_triangulation_logo(
    filename="lucid_triangulation_logo_fixed.svg",
):
    size = 1000
    dwg = svgwrite.Drawing(filename, size=(size, size), profile="full")

    # Colors
    COLOR_BG = "#0D0E12"
    COLOR_RED = "#E63946"
    COLOR_BLUE = "#005596"
    COLOR_PURPLE = "#6A0DAD"
    COLOR_WHITE = "#FFFFFF"

    dwg.add(dwg.rect(insert=(0, 0), size=(size, size), fill=COLOR_BG))

    cx, cy = 500, 380
    r = 150
    stroke_width = 14
    offset_distance = 45  # Spreads the 3 triangles outward from the center

    def get_triangle_vertices(center_x, center_y, radius, angle_offset_deg):
        points = []
        for i in range(3):
            angle_rad = math.radians(angle_offset_deg + i * 120)
            x = center_x + radius * math.cos(angle_rad)
            y = center_y + radius * math.sin(angle_rad)
            points.append((x, y))
        return points

    # Calculate offset centers for the 3 triangles so all 3 colors are visible
    centers = [
        (
            cx + offset_distance * math.cos(math.radians(-90)),
            cy + offset_distance * math.sin(math.radians(-90)),
        ),  # Top
        (
            cx + offset_distance * math.cos(math.radians(30)),
            cy + offset_distance * math.sin(math.radians(30)),
        ),  # Bottom Right
        (
            cx + offset_distance * math.cos(math.radians(150)),
            cy + offset_distance * math.sin(math.radians(150)),
        ),  # Bottom Left
    ]

    triangles_data = [
        (get_triangle_vertices(centers[0][0], centers[0][1], r, -90), COLOR_RED),
        (
            get_triangle_vertices(centers[1][0], centers[1][1], r, -90),
            COLOR_BLUE,
        ),
        (
            get_triangle_vertices(centers[2][0], centers[2][1], r, -90),
            COLOR_PURPLE,
        ),
    ]

    logo_mark = dwg.g(id="geometric_crest")

    # Draw the 3 distinct vector triangles
    for points, color in triangles_data:
        logo_mark.add(
            dwg.polygon(
                points=points,
                fill="none",
                stroke=color,
                stroke_width=stroke_width,
                stroke_linejoin="round",
                opacity=0.85,
            )
        )

    # Sharp white core in the exact mathematical center
    inner_core = get_triangle_vertices(cx, cy - 10, r * 0.22, -90)
    logo_mark.add(
        dwg.polygon(
            points=inner_core,
            fill=COLOR_WHITE,
            opacity=0.95,
        )
    )

    dwg.add(logo_mark)

    # Swiss Typography Alignment
    dwg.add(
        dwg.text(
            "LUCID TRIANGULATION",
            insert=(cx, 710),
            text_anchor="middle",
            fill=COLOR_WHITE,
            font_size="38px",
            font_family="Montserrat, Helvetica, Arial, sans-serif",
            font_weight="bold",
            letter_spacing="12px",
        )
    )

    dwg.add(
        dwg.text(
            "RECORDS",
            insert=(cx, 765),
            text_anchor="middle",
            fill="#ADAFAF",
            font_size="22px",
            font_family="Montserrat, Helvetica, Arial, sans-serif",
            font_weight="300",
            letter_spacing="18px",
        )
    )

    dwg.save()
    print(f"✅ Generated: {filename}")
    files.download(filename)


create_lucid_triangulation_logo()
```
